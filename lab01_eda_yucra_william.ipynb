5.5. Construir un resumen de columnas


```python
resumen_columnas = pd.DataFrame({
"tipo_pandas": df.dtypes.astype(str),
"datos_presentes": df.notna().sum(),
"datos_nulos": df.isna().sum(),
"valores_unicos": df.nunique(dropna=True)
})
resumen_columnas
```

Separa las columnas que pandas reconoce como numéricas y categóricas:


```python
numeric_cols_to_convert = ['Record_ID', 'Confirmed', 'Recovered', 'Deaths', 'Active', 'Tests_Conducted', 'Vaccination_Rate_%']
for col in numeric_cols_to_convert:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

columnas_numericas = df.select_dtypes(
    include=np.number
).columns.tolist()
columnas_categoricas = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()
print("Columnas numericas:", columnas_numericas)
print("Columnas categoricas:", columnas_categoricas)
```

6. Paso 4 · Evaluar la calidad inicial
En este paso sólo diagnosticarás. No elimines ni reemplaces datos todavía.

   6.1. Valores nulos


```python
resumen_nulos = pd.DataFrame({
"cantidad_nulos": df.isnull().sum(),
"porcentaje_nulos": (df.isnull().mean() *
100).round(2)
})
resumen_nulos = resumen_nulos.sort_values(
by="porcentaje_nulos",
ascending=False
)
resumen_nulos
```

Para mostrar sólo las columnas que tienen al menos un nulo:

```python
columnas_con_nulos = resumen_nulos[
resumen_nulos["cantidad_nulos"] > 0
]
columnas_con_nulos
```

6.2. Filas duplicadas


```python
cantidad_duplicados = df.duplicated().sum()
porcentaje_duplicados = cantidad_duplicados / len(df) *100
print("Filas duplicadas:", cantidad_duplicados)
print("Porcentaje duplicado:", round(porcentaje_duplicados, 2), " %")
```

6.3. Cantidad de valores diferentes

```python
valores_diferentes = df.nunique(dropna=False).sort_values()
valores_diferentes

```

7. Paso 5 · Estadística descriptiva y percentiles

  7.1. Resumen de columnas numéricas

```python
print(columnas_numericas)
```

Luego genera el resumen:


```python
if columnas_numericas:
   resumen_numerico = df[columnas_numericas].describe().T
   display(resumen_numerico)
else:
   print("El dataset no tiene columnas numericas detectadas")
```

7.2. Comparar media y mediana

```python
if columnas_numericas:
  comparacion_centro = pd.DataFrame({
    "media": df[columnas_numericas].mean(),
    "mediana": df[columnas_numericas].median(),
    "desviacion_estandar": df[columnas_numericas].std(),
    "minimo": df[columnas_numericas].min(),
  "maximo": df[columnas_numericas].max()
  })
  display(comparacion_centro)
```

7.3. Calcular percentiles ampliados

```python
if columnas_numericas:
  tabla_percentiles = df[columnas_numericas].quantile(
    [0.00, 0.05, 0.25, 0.50, 0.75, 0.95, 1.00]
).T
  tabla_percentiles.columns = [
    "min", "p05", "p25", "p50_mediana",
    "p75", "p95", "max"
]
  display(tabla_percentiles)
```

7.4. Resumen de columnas categóricas


```python
if columnas_categoricas:
  resumen_categorico = df[columnas_categoricas].describe().T
  display(resumen_categorico)
else:
  print("No se detectaron columnas categoricas")
```

7.5. Frecuencias de una categoría

```python
print(columnas_categoricas)
# Cambia el indice si la primera columna no es adecuada.
columna_categorica = columnas_categoricas[0]
print("Columna elegida:", columna_categorica)
```

Calcula cantidad y porcentaje:

```python
frecuencia = df[columna_categorica].value_counts(
    dropna=False
)
porcentaje = df[columna_categorica].value_counts(
    dropna=False,
    normalize=True
).mul(100).round(2)
tabla_frecuencias = pd.DataFrame({
    "cantidad": frecuencia,
    "porcentaje": porcentaje
})
tabla_frecuencias.head(15)
```

8. Paso 6 · Visualizar las distribuciones
Cada gráfico debe tener título, nombre de ejes y una explicación. El gráfico no reemplaza la estadística descriptiva; ambos se
complementan.

  8.1. Elegir una columna numérica


```python
print(columnas_numericas)
# Cambia el indice para elegir una variable significativa.
columna_numerica = columnas_numericas[0]
print("Columna elegida:", columna_numerica)
```

8.2. Histograma con media y mediana

```python
media = df[columna_numerica].mean()
mediana = df[columna_numerica].median()
plt.figure(figsize=(10, 5))
sns.histplot(
    data=df,
    x=columna_numerica,
    bins=20,
    kde=True,
    color="#0068B3"
)
plt.axvline(
    media,
    color="#C0392B",
    linestyle="--",
    label=f"Media: {media:.2f}"
)
plt.axvline(
    mediana,
    color="#007A6A",
    linestyle="-.",
    label=f"Mediana: {mediana:.2f}"
)
plt.title(f"Distribucion de {columna_numerica}")
plt.xlabel(columna_numerica)
plt.ylabel("Cantidad de registros")
plt.legend()
plt.show()
```

8.3. Diagrama de caja


```python
plt.figure(figsize=(10, 3.5))
sns.boxplot(
    x=df[columna_numerica],
    color="#8ED1C6"
)
plt.title(f"Diagrama de caja de {columna_numerica}")
plt.xlabel(columna_numerica)
plt.show()
```

8.4. Gráfico de barras para una columna categórica

```python
top_categorias = df[columna_categorica].value_counts(
    dropna=False
).head(10)
plt.figure(figsize=(10, 5))
sns.barplot(
    x=top_categorias.values,
    y=top_categorias.index.astype(str),
    color="#009C88"
)
plt.title(f"Categorias mas frecuentes de {columna_categorica}")
plt.xlabel("Cantidad de registros")
plt.ylabel(columna_categorica)
plt.tight_layout()
plt.show()
```

8.5. Vista general de hasta seis variables numéricas (opcional)

```python
seleccion_numericas = columnas_numericas[:6]
if seleccion_numericas:
  df[seleccion_numericas].hist(
    bins=20,
    figsize=(12, 8),
    color="#0068B3",
    edgecolor="white"
)
plt.suptitle("Distribuciones de variables numericas")
plt.tight_layout()
plt.show()
```

8.6. Exploración opcional de relación entre variables

```python
seleccion_correlacion = columnas_numericas[:8]
if len(seleccion_correlacion) >= 2:
  matriz_correlacion = df[seleccion_correlacion].corr()
  plt.figure(figsize=(9, 7))
  sns.heatmap(
    matriz_correlacion,
    annot=True,
    fmt=".2f",
    cmap="vlag",
    center=0
)
plt.title("Correlacion entre variables numericas")
plt.tight_layout()
plt.show()
```

## 1. Business Understanding

El objetivo es analizar los salarios de los trabajadores del sector Datos/IT en diferentes países. La pregunta inicial del Laboratorio 01 fue: ¿Cuál es el salario promedio de los trabajadores del sector Datos/IT. A partir de esta pregunta, se busca entender cómo se distribuyen los salarios y cómo pueden cambiar según factores como el nivel de experiencia, el cargo, la ubicación geográfica y la modalidad de trabajo. El análisis ayudará a tener una visión clara de las remuneraciones en este sector.

## 2. Data Understanding

El dataset tiene 8.002 registros y 11 columnas. Cada fila corresponde a un salario de un trabajador del sector Datos/IT. Las variables incluyen año, nivel de experiencia, tipo de empleo, cargo, salario, moneda, salario en dólares, residencia, trabajo remoto, ubicación de la empresa y tamaño de la empresa. No se encontraron valores vacíos en el EDA. Sin embargo, se identificaron 3.299 filas repetidas, lo que representa el 41,23%. Además, la variable salary_in_usd tiene una media de 149.409,52 USD y una mediana de 142.200 USD.

## 3. Data Preparation

Antes de comenzar cualquier modelado futuro, es indispensable revisar y corregir las **3.299 filas duplicadas** que se han encontrado en el EDA. No se hallaron valores nulos, así que no será necesario imputar datos por esa razón. También conviene revisar los valores extremos, en especial en las variables salariales, porque `salary` muestra cifras muy altas. Para comparar de forma adecuada las remuneraciones entre países, emplearemos principalmente `salary_in_usd`, pues permite trabajar con salarios expresados en una sola moneda.

## 4. Modeling

Se propone usar una técnica de regresión para modelar los datos, con `salary_in_usd` como variable objetivo. Entre las variables que podrían influir en el salario se encuentran `experience_level`, `job_title`, `employment_type`, `remote_ratio`, `company_location` y `company_size`. Las variables categóricas necesitarán ser transformadas antes de poder incluirse en el modelo. Esta parte del trabajo se llevará a cabo en laboratorios posteriores, por lo que no es necesario ejecutar el modelo en esta actividad.

## 5. Evaluation

Para evaluar un futuro modelo de regresión se podrían usar métricas como **MAE** y **RMSE**, que medirían qué tan grande es el error al estimar el salario en dólares. También es necesario separar los datos en entrenamiento y prueba para comprobar el desempeño del modelo con información que no haya usado el entrenamiento. El modelo debe mostrar un error suficientemente bajo para considerarse útil para el análisis de los salarios del sector Datos/IT.


## 6. Deployment

Los resultados podrían servir para mostrar datos sobre los salarios del sector de Datos/IT a través de un informe o un panel de control. Por ejemplo, se podrían hacer comparaciones de salarios según el nivel de experiencia, el cargo, el país, la forma de trabajo o el tamaño de la empresa. Esto ayudaría a convertir los hallazgos del análisis en información clara y sencilla de entender. Esta información podría usarse para comprender mejor las remuneraciones del sector tecnológico.
